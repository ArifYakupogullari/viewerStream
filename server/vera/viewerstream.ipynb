# viewerStream: remote Gaussian-splat renderer on Vera

Run this inside a **Vera OnDemand Jupyter session** (https://vera.c3se.chalmers.se) with **one A100**
(the renderer needs a GPU; the session's time limit is the renderer's maximum lifetime).

1. One-time only: open a terminal in this session and run `bash server/vera/setup_env.sh` from the repo root.
2. Set `PLY` below to a trained splat (`.../point_cloud/iteration_N/point_cloud.ply`), or use `backend="fake"` for a GPU-free test pattern.
3. Run the start cell, then open the link (laptop, or scan the QR code on a phone on eduroam / Chalmers VPN),
   or paste it into the StreamBIM widget's **Remote** panel and press **Connect**.

Viewers need to log in to the Vera portal with a Chalmers account. The link only works while the renderer runs.

```python
import sys
from pathlib import Path

REPO = Path("/cephyr/users/ahmetari/Vera/viewerStream")   # where this repo is cloned on Vera
sys.path.insert(0, str(REPO / "server" / "vera"))
import launch

# Trained splat to serve (graphdeco layout, from final_hope.sh). Find candidates with:
#   find /cephyr/NOBACKUP/groups/c3se2026-1-14 -path '*point_cloud/iteration_*' -name point_cloud.ply
PLY = "/cephyr/NOBACKUP/groups/c3se2026-1-14/gs-june/point_cloud/iteration_75000/point_cloud.ply"
```

## Start

```python
session = launch.start(ply=PLY)          # launch.start(backend="fake") for the test pattern
session.show()
```

## Status

```python
session.status()
print(session.log_tail())
```
